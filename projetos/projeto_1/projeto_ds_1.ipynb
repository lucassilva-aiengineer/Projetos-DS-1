# Qual o tamanho da frota de carros elétricos de washington?
# Qual a fatia do mercado que a Tesla possui?

```python
tamanho_frota = contagem_marcas['count'].sum()
qntd_tesla = contagem_marcas.at[0, 'count']

parcela_tesla = (qntd_tesla / tamanho_frota) * 100
```

```python
print(f"Parcela Tesla: {parcela_tesla:.2f}%")
```

#Qual o tamanho da Frota nacional de carros elétricos?

```python
print(f"A minha estimatica de quantidade de veículos elétricos que os EUA possuem: {tamanho_frota * 50:,} de unidades")
```
